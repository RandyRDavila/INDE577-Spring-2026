# Ensemble Methods: Learning as a Team

A single model can be unstable or systematically biased. An **ensemble** combines several models so their individual mistakes are less influential.

## Learning goals

- Explain why diverse models can outperform one model.
- Distinguish parallel methods such as bagging from sequential methods such as boosting.
- Compare a decision tree, random forest, and AdaBoost classifier through one consistent API.

## Two ensemble strategies

**Bagging** trains models independently on perturbed data and combines their votes. It primarily reduces variance. Random forests add random feature selection to bagging.

**Boosting** trains models in sequence. Each new model concentrates on examples the current ensemble gets wrong. It primarily reduces bias, but can become sensitive to noise.

```python
from pathlib import Path
import sys

import numpy as np

repo_root = next(
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "src" / "rice_ml").is_dir()
)
sys.path.insert(0, str(repo_root / "src"))

from rice_ml.supervised_ml import (
    AdaBoostClassifier,
    DecisionTreeClassifier,
    RandomForestClassifier,
)
```

```python
rng = np.random.default_rng(11)
X = rng.normal(size=(180, 2))
y = np.where(X[:, 0] ** 2 + X[:, 1] ** 2 > 1.0, "outer", "inner")

order = rng.permutation(len(X))
train, test = order[:120], order[120:]
X_train, X_test = X[train], X[test]
y_train, y_test = y[train], y[test]

models = {
    "single tree": DecisionTreeClassifier(max_depth=3),
    "random forest": RandomForestClassifier(
        n_estimators=75, max_depth=4, random_state=11
    ),
    "AdaBoost": AdaBoostClassifier(n_estimators=40, learning_rate=0.7),
}

for name, model in models.items():
    model.fit(X_train, y_train)
    accuracy = np.mean(model.predict(X_test) == y_test)
    print(f"{name:14s} accuracy = {accuracy:.3f}")
```

## Reading the comparison

Do not expect an ensemble to win every split of every dataset. The useful question is *why* performance changes. The forest averages many high-variance trees. AdaBoost combines simple boundaries into a more expressive boundary.
